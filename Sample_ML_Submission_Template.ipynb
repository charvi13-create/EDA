## 3. ***Data Wrangling***

### Data Wrangling Code

````python
df_titles_clean = df_titles.copy()
df_credits_clean = df_credits.copy()
df_titles_clean.drop_duplicates(inplace=True)
df_credits_clean.drop_duplicates(inplace=True)

for column in df_titles_clean.select_dtypes(include='object').columns:
    df_titles_clean[column] = df_titles_clean[column].str.strip()
for column in df_credits_clean.select_dtypes(include='object').columns:
    df_credits_clean[column] = df_credits_clean[column].str.strip()

numeric_columns = [
    'release_year', 'runtime', 'seasons',
    'imdb_score', 'imdb_votes',
    'tmdb_popularity', 'tmdb_score'
]

for column in numeric_columns:
    df_titles_clean[column] = pd.to_numeric(
        df_titles_clean[column], errors='coerce'
    )
print("Data Wrangling Completed Successfully!")
print("\nCleaned Titles Dataset Shape:", df_titles_clean.shape)
print("Cleaned Credits Dataset Shape:", df_credits_clean.shape)
print("\nMissing Values in Cleaned Titles Dataset:")
print(df_titles_clean.isnull().sum())
print("\nMissing Values in Cleaned Credits Dataset:")
print(df_credits_clean.isnull().sum())
````

### What all manipulations have you done and insights you found?

1. **Creating Dataset Copies:** Copies of the original titles and credits datasets were created to preserve the original data during the cleaning process.

2. **Removing Duplicate Records:** Duplicate rows were identified and removed from both datasets to avoid repeated records affecting the analysis.
3. **Handling Text Data:** Extra spaces were removed from text-based columns to maintain consistency and improve data quality.
4. **Converting Data Types:** Numerical attributes such as release year, runtime, seasons, IMDb score, IMDb votes, TMDB popularity and TMDB score were converted into appropriate numerical data types. Invalid values were treated as missing values.
5. **Checking Missing Values:** Missing values were examined after the cleaning operations to understand the remaining data quality issues.

These manipulations helped improve the consistency and organisation of both datasets. The cleaned data is now better prepared for exploratory analysis, feature engineering and Regression model development. Missing values will be handled using suitable preprocessing techniques in the later stages of the project.

#### Chart - 6

````python
import ast
def extract_genres(value):
    if pd.isna(value):
        return []
    try:
        parsed = ast.literal_eval(value) if isinstance(value, str) else value
        return parsed if isinstance(parsed, list) else [str(parsed)]
    except (ValueError, SyntaxError):
        return [str(value)]
genre_counts = (
    df_titles_clean['genres']
    .apply(extract_genres)
    .explode()
    .dropna()
    .value_counts()
    .head(10)
    .reset_index()
)
genre_counts.columns = ['Genre', 'Number of Titles']

plt.figure(figsize=(11, 6))
sns.barplot(
    data=genre_counts,
    x='Number of Titles',
    y='Genre',
    hue='Genre',
    palette='viridis',
    legend=False
)

plt.title('Top 10 Genres on Amazon Prime Video')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
````

A bar chart was selected to compare the number of titles across different genres. It makes it easy to identify the most and least represented genres on Amazon Prime Video.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

In this project, three hypotheses are formulated based on the analysis of the Amazon Prime Video dataset. Statistical tests are performed to examine the differences and relationships between IMDb ratings, content types, audience votes, and release years. The results are used to determine whether the hypotheses are accepted or rejected based on a 5% significance level.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H0):** There is no significant difference between the average IMDb ratings of movies and TV shows on Amazon Prime Video.

**Alternative Hypothesis (H1):** There is a significant difference between the average IMDb ratings of movies and TV shows on Amazon Prime Video.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import ttest_ind

movies = df_titles_clean[
    (df_titles_clean['type'] == 'MOVIE') &
    (df_titles_clean['imdb_score'].notna())
]['imdb_score']
shows = df_titles_clean[
    (df_titles_clean['type'] == 'SHOW') &
    (df_titles_clean['imdb_score'].notna())
]['imdb_score']
t_stat, p_value = ttest_ind(
    movies,
    shows,
    equal_var=False
)
print("T-statistic:", t_stat)
print("P-value:", p_value)
````

##### Which statistical test have you done to obtain P-Value?

Welch's Independent Two-Sample T-Test.

##### Why did you choose the specific statistical test?

Welch's Independent Two-Sample T-Test was chosen because the objective is to compare the average IMDb ratings of two independent groups: movies and TV shows. It is suitable for comparing the means of two groups without assuming equal population variances.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H0):** There is no significant relationship between the number of IMDb votes and IMDb ratings of Amazon Prime Video titles.

**Alternative Hypothesis (H1):** There is a significant relationship between the number of IMDb votes and IMDb ratings of Amazon Prime Video titles.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import spearmanr
correlation_data = df_titles_clean.dropna(
    subset=['imdb_votes', 'imdb_score']
)
correlation, p_value = spearmanr(
    correlation_data['imdb_votes'],
    correlation_data['imdb_score']
)
print("Spearman Correlation:", correlation)
print("P-value:", p_value)
if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("There is a significant relationship between IMDb votes and ratings.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is no significant relationship between IMDb votes and ratings.")
````

##### Which statistical test have you done to obtain P-Value?

Spearman's Rank Correlation Test.

##### Why did you choose the specific statistical test?

Spearman's Rank Correlation Test was chosen because the objective is to identify the relationship between IMDb votes and IMDb ratings. It is suitable for measuring the strength and direction of a monotonic relationship between two numerical variables and does not require the data to follow a normal distribution.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H0):** There is no significant difference between the average IMDb ratings of titles released before 2000 and titles released from 2000 onwards.

**Alternative Hypothesis (H1):** There is a significant difference between the average IMDb ratings of titles released before 2000 and titles released from 2000 onwards.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import ttest_ind
release_data = df_titles_clean.dropna(
    subset=['release_year', 'imdb_score']
)
before_2000 = release_data[
    release_data['release_year'] < 2000
]['imdb_score']
from_2000 = release_data[
    release_data['release_year'] >= 2000
]['imdb_score']
t_stat, p_value = ttest_ind(
    before_2000,
    from_2000,
    equal_var=False
)
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("There is a significant difference in average IMDb ratings.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is no significant difference in average IMDb ratings.")
````

Welch's Independent Two-Sample T-Test.

### 2. Handling Outliers

````python
df_model = df_titles_clean.copy()

numeric_columns = df_model.select_dtypes(include=['number']).columns
categorical_columns = df_model.select_dtypes(include=['object']).columns
for column in numeric_columns:
    if column != 'imdb_score':
        df_model[column] = df_model[column].fillna(
            df_model[column].median()
        )
for column in categorical_columns:
    mode_value = df_model[column].mode()
    if not mode_value.empty:
        df_model[column] = df_model[column].fillna(mode_value[0])
df_model = df_model.dropna(subset=['imdb_score'])
outlier_columns = [
    'runtime',
    'seasons',
    'imdb_votes',
    'tmdb_popularity'
]

for column in outlier_columns:
    Q1 = df_model[column].quantile(0.25)
    Q3 = df_model[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df_model[column] = df_model[column].clip(
        lower=lower_limit,
        upper=upper_limit
    )

print("Missing values handled successfully!")
print("Outlier treatment completed successfully!")
print("Final Dataset Shape:", df_model.shape)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

I used the Interquartile Range (IQR) method to identify outliers in numerical variables such as runtime, seasons, IMDb votes, and TMDB popularity. I applied the capping technique to replace extreme values with the calculated lower and upper limits instead of removing entire rows. This helps reduce the influence of extreme values while preserving important observations in the dataset.

#### 1. Expand Contraction

````python
import re

df_text = df_titles_clean[['description']].copy()
df_text['description'] = df_text['description'].fillna('')
contractions_dict = {
    "don't": "do not",
    "doesn't": "does not",
    "didn't": "did not",
    "isn't": "is not",
    "aren't": "are not",
    "wasn't": "was not",
    "weren't": "were not",
    "can't": "cannot",
    "couldn't": "could not",
    "wouldn't": "would not",
    "shouldn't": "should not",
    "it's": "it is",
    "i'm": "i am",
    "they're": "they are",
    "we're": "we are",
    "you're": "you are"
}

def expand_contractions(text):
    for contraction, expansion in contractions_dict.items():
        text = re.sub(
            r"\b" + re.escape(contraction) + r"\b",
            expansion,
            text,
            flags=re.IGNORECASE
        )
    return text

df_text['description'] = df_text['description'].apply(expand_contractions)

print(df_text['description'].head())
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
ridge_model = Ridge()


param_grid = {
    'alpha': [0.01, 0.1, 1, 10, 100]
}


grid_search = GridSearchCV(
    estimator=ridge_model,
    param_grid=param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1
)

grid_search.fit(X_train_scaled, y_train)

best_ridge_model = grid_search.best_estimator_

y_pred_ridge = best_ridge_model.predict(X_test_scaled)

ridge_mae = mean_absolute_error(y_test, y_pred_ridge)
ridge_mse = mean_squared_error(y_test, y_pred_ridge)
ridge_rmse = np.sqrt(ridge_mse)
ridge_r2 = r2_score(y_test, y_pred_ridge)

print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validation R2 Score:", grid_search.best_score_)

print("\nTuned Ridge Regression Performance")
print("MAE:", ridge_mae)
print("MSE:", ridge_mse)
print("RMSE:", ridge_rmse)
print("R2 Score:", ridge_r2)
````

#### 3. Explain each evaluation metric's indication towards business and the business impact pf the ML model used.

**MAE (Mean Absolute Error):** Measures the average difference between actual and predicted IMDb ratings. A lower MAE indicates more accurate predictions, helping Amazon Prime Video estimate audience ratings more reliably.

**MSE (Mean Squared Error):** Measures the average squared prediction error. A lower MSE indicates fewer large prediction errors, helping reduce inaccurate rating predictions.

**RMSE (Root Mean Squared Error):** Measures prediction errors in the same unit as IMDb ratings. A lower RMSE indicates better prediction accuracy and helps the platform understand the typical prediction error.

**R² Score:** Measures how much variation in IMDb ratings is explained by the model. A higher R² Score indicates that the model explains more variation in ratings.

**Business Impact:** The Random Forest Regression model can help Amazon Prime Video identify patterns in content ratings, support content analysis, and improve data-driven content recommendations. However, predicted IMDb ratings should be used as supporting information rather than the only factor in content acquisition and investment decisions.
